```python
# Q48. For two given sets, display their:
# union, intersection, difference (A - B), and symmetric difference.

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union =", A | B)
print("Intersection =", A & B)
print("Difference =", A - B)
print("Symmetric Difference =", A ^ B)
```

```python
# Q51. Write a program that checks whether two given sets
# have no common elements.

A = {1, 2, 3}
B = {4, 5, 6}

if A.isdisjoint(B):
    print("No common elements")
else:
    print("Common elements exist")
```

```python
# Q52. Given two sets A and B, print whether A is a subset of B
# and whether A is a superset of B.

A = {1, 2}
B = {1, 2, 3, 4}

print("A is subset of B:", A.issubset(B))
print("A is superset of B:", A.issuperset(B))
```
