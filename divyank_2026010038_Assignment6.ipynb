# Part A: Normal Distribution

A normal distribution is used for continuous data and is described by its mean and standard deviation.

## Question 1: Student Marks

The marks of students follow a normal distribution with a mean of 70 and a standard deviation of 10. Calculate the probability that a randomly selected student scores less than 80 marks.

```python
from scipy.stats import norm

mean = 70
std_dev = 10
marks = 80

probability = norm.cdf(marks, mean, std_dev)

print("Probability of scoring less than 80:", probability)
print("Percentage:", probability * 100)
```

**Explanation:** `norm.cdf()` gives the cumulative probability up to the specified value. It calculates \(P(X \leq 80)\). The probability is approximately **0.8413**, or **84.13%**.

## Question 2: Machine Processing Time

The processing time of a machine follows a normal distribution with a mean of 50 seconds and a standard deviation of 5 seconds. Calculate the probability that processing time lies between 45 and 55 seconds.

```python
from scipy.stats import norm

mean = 50
std_dev = 5
lower_limit = 45
upper_limit = 55

p1 = norm.cdf(lower_limit, mean, std_dev)
p2 = norm.cdf(upper_limit, mean, std_dev)
probability = p2 - p1

print("Probability between 45 and 55 seconds:", probability)
print("Percentage:", probability * 100)
```

**Explanation:** Subtract the cumulative probability at the lower limit from that at the upper limit:

\[P(45 < X < 55) = F(55) - F(45)\]

The result is approximately **0.6827**, or **68.27%**.

# Part B: Binomial Distribution

A binomial distribution models a fixed number of independent trials, each with two possible outcomes and a constant probability of success.

## Question 3: Quality Check of Products

The probability that a product is defective is 0.1. If 5 products are selected independently, calculate the probability that exactly 2 are defective.

```python
from scipy.stats import binom

n = 5
p = 0.1
x = 2

probability = binom.pmf(x, n, p)

print("Probability of exactly 2 defective products:", probability)
```

**Explanation:** `n = 5` is the number of products, `p = 0.1` is the probability of a defective product, and `x = 2` is the required number. `binom.pmf()` calculates the probability of exactly 2 defective products: **0.0729**, or **7.29%**.

## Question 4: Students Passing an Examination

The probability that a student passes an examination is 0.8. If 6 students are selected independently, calculate the probability that at least 5 pass.

```python
from scipy.stats import binom

n = 6
p = 0.8

p_less_than_5 = binom.cdf(4, n, p)
probability = 1 - p_less_than_5

print("Probability that at least 5 students pass:", probability)
```

**Explanation:** “At least 5” means 5 or 6 students pass. Use the complement:

\[P(X \geq 5) = 1 - P(X \leq 4)\]

The result is **0.65536**, or **65.536%**.

# Part C: Chi-Square Distribution

A chi-square goodness-of-fit test compares observed frequencies with expected frequencies. The null hypothesis here states that all four snack types are equally preferred.

## Question 5: Preference for Four Types of Snacks

A survey records the preferences of 100 students for four types of snacks. If all four snacks are equally preferred, the expected frequency is 25 for each snack. Write a Python program to calculate the chi-square statistic and p-value, calculate degrees of freedom, and make a decision at the 5% significance level using the critical value 7.815.

| Snack | Observed frequency | Expected frequency |
|---|---:|---:|
| Chips | 30 | 25 |
| Biscuits | 20 | 25 |
| Sandwiches | 25 | 25 |
| Fruits | 25 | 25 |
| **Total** | **100** | **100** |

```python
from scipy.stats import chisquare

observed = [30, 20, 25, 25]
expected = [25, 25, 25, 25]

chi_stat, p_value = chisquare(
    f_obs=observed,
    f_exp=expected
)

df = len(observed) - 1
critical_value = 7.815
alpha = 0.05

print("Chi-square statistic:", chi_stat)
print("P-value:", p_value)
print("Degrees of freedom:", df)

if chi_stat > critical_value:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

**Explanation:**

The chi-square statistic is:

\[\chi^2 = \sum \frac{(O-E)^2}{E}\]

where \(O\) is the observed frequency and \(E\) is the expected frequency.

Degrees of freedom:

\[df = k - 1 = 4 - 1 = 3\]

The statistic is **2.0**, less than the critical value **7.815**. The p-value is approximately **0.5724**, greater than 0.05. Therefore, **fail to reject the null hypothesis**. There is insufficient evidence at the 5% significance level that snack preferences differ from an equal distribution.

# Part D: Poisson Distribution

A Poisson distribution calculates the probability of a number of events occurring within a fixed interval when events occur independently at a constant average rate.

## Question 6: Customer Arrivals at a Shop

A shop receives an average of 4 customers every hour. Calculate the probability that exactly 2 customers arrive in one hour.

```python
from scipy.stats import poisson

mean = 4
x = 2

probability = poisson.pmf(x, mean)

print("Probability of exactly 2 customers:", probability)
print("Percentage:", probability * 100)
```

**Explanation:** `mean = 4` is the average number of customer arrivals per hour. `poisson.pmf()` calculates the probability of exactly 2 arrivals: approximately **0.1465**, or **14.65%**.

## Question 7: Errors in a Computer Program

A computer program generates an average of 3 errors per day. Calculate the probability that at most 2 errors occur in a day.

```python
from scipy.stats import poisson

mean = 3
maximum_errors = 2

probability = poisson.cdf(maximum_errors, mean)

print("Probability of at most 2 errors:", probability)
print("Percentage:", probability * 100)
```

**Explanation:** “At most 2 errors” means 0, 1 or 2 errors. `poisson.cdf(2, 3)` adds the probabilities of these outcomes: approximately **0.4232**, or **42.32%**.

# Quick Reference: Python Functions Used

| Normal | `norm.cdf(b, mean, std_dev) - norm.cdf(a, mean, std_dev)` | Probability between a and b |
| Binomial | `binom.pmf(x, n, p)` | Probability of exactly x successes |
| Binomial | `binom.cdf(x, n, p)` | Probability of at most x successes |
| Chi-square | `chisquare(observed, expected)` | Chi-square statistic and p-value |
| Poisson | `poisson.pmf(x, mean)` | Probability of exactly x events |
| Poisson | `poisson.cdf(x, mean)` | Probability of at most x events |
