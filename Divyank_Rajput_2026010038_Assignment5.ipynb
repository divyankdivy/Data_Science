# Part A: Basic DataFrame Operations
## Question 1: Import and Display Dataset

```python
import pandas as pd

df = pd.read_csv("student_performance.csv")
print("Complete DataFrame:")
display(df)
```

```python
print("First 5 rows:")
display(df.head(5))
print("Last 3 rows:")
display(df.tail(3))
```

## Question 2: Basic Information of DataFrame

```python
print("info():")
df.info()
print("\nshape   :", df.shape)
print("size    :", df.size)
print("columns :", list(df.columns))
print("\ndtypes:")
print(df.dtypes)
```

**Explanation**

| Function | Information provided |
|---|---|
| `info()` | Concise summary: index range, column names, non-null counts, data types and memory usage |
| `shape` | Tuple of (number of rows, number of columns) |
| `size` | Total number of elements (rows x columns) |
| `columns` | Index object containing the column names |
| `dtypes` | Data type of each column |

## Question 3: Create and Modify DataFrame

```python
subjects = ["Python_Marks", "AI_Marks", "Network_Marks"]

df["Total_Marks"] = df[subjects].sum(axis=1)
df["Average_Marks"] = df[subjects].mean(axis=1)
df["Result"] = df["Average_Marks"].apply(lambda x: "Pass" if x >= 40 else "Fail")

display(df)
```

## Question 4: Adding and Deleting Columns

```python
def performance(avg):
    if avg >= 85:
        return "Excellent"
    elif avg >= 70:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance_Level"] = df["Average_Marks"].apply(performance)
df = df.drop(columns="Age")

display(df)
```

---
# Part B: Mathematical and Statistical Functions
## Question 5: Statistical Analysis

```python
stats = pd.DataFrame({
    "Sum": df[subjects].sum(),
    "Mean": df[subjects].mean(),
    "Min": df[subjects].min(),
    "Max": df[subjects].max(),
    "Median": df[subjects].median()
})
display(stats)
```

## Question 6: Aggregate Function - agg()

```python
agg_result = df[subjects].agg(["min", "max", "mean", "std"])
agg_result.index = ["Minimum", "Maximum", "Average", "Standard Deviation"]
display(agg_result)
```

## Question 7: Percentage Change

```python
df = df.sort_values("Student_ID")
df["Python_Pct_Change"] = df["Python_Marks"].pct_change() * 100

display(df[["Student_ID", "Python_Marks", "Python_Pct_Change"]])
```

# Part C: Data Sorting
## Question 8: Sorting Using sort_values()

```python
print("Python_Marks in ascending order:")
display(df.sort_values("Python_Marks", ascending=True))
```

```python
print("Average_Marks in descending order:")
display(df.sort_values("Average_Marks", ascending=False))
```

```python
print("Department ascending, Python_Marks descending:")
display(df.sort_values(["Department", "Python_Marks"], ascending=[True, False]))
```

```python
print("Top 3 students by Average_Marks:")
display(df.sort_values("Average_Marks", ascending=False).head(3))
```

## Question 9: Sorting Using sort_index()

```python
df_idx = df.set_index("Student_ID")
print("Student_ID set as index:")
display(df_idx)
```

```python
df_idx = df_idx.sort_index(ascending=False)
print("Sorted by index in descending order:")
display(df_idx)
```

```python
df_idx = df_idx.reset_index()
print("After resetting the index:")
display(df_idx)
```

# Part D: Data Extraction and Filtering
## Question 10: Data Extraction Using Relational Operators

```python
print("Python_Marks > 80:")
display(df[df["Python_Marks"] > 80])

print("Attendance >= 90:")
display(df[df["Attendance"] >= 90])

print("AI_Marks < 70:")
display(df[df["AI_Marks"] < 70])

print("Network_Marks != 75:")
display(df[df["Network_Marks"] != 75])
```

## Question 11: Data Extraction Using Logical Operators

```python
print("Python_Marks > 75 AND AI_Marks > 80:")
display(df[(df["Python_Marks"] > 75) & (df["AI_Marks"] > 80)])

print("Attendance > 90 OR Study_Hours > 5:")
display(df[(df["Attendance"] > 90) | (df["Study_Hours"] > 5)])

print("CSE students with Average_Marks > 75:")
display(df[(df["Department"] == "CSE") & (df["Average_Marks"] > 75)])

print("Students not in the IT department:")
display(df[~(df["Department"] == "IT")])
```

## Question 12: Data Extraction Using iloc
*Note: rows 3 to 7 and columns 2 to 5 are taken as 1-based positions, i.e. `iloc[2:7, 1:5]`.*

```python
print("First 5 rows:")
display(df.iloc[:5])

print("Rows 3 to 7 and columns 2 to 5:")
display(df.iloc[2:7, 1:5])

print("Last 4 rows and first 3 columns:")
display(df.iloc[-4:, :3])

print("Second row, Python_Marks column:")
print(df.iloc[1, df.columns.get_loc("Python_Marks")])
```

# Part E: Data Visualization
## Question 13: Line Plot

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(df["Student_ID"], df["Python_Marks"], marker="o", color="blue")
plt.title("Python Marks of Students")
plt.xlabel("Student_ID")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

## Question 14: Bar Chart

```python
mean_marks = df[subjects].mean()
print(mean_marks)

plt.figure(figsize=(7, 5))
plt.bar(["Python", "AI", "Network Programming"], mean_marks, color=["steelblue", "orange", "green"])
plt.title("Average Marks by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.show()
```

## Question 15: Histogram

```python
plt.figure(figsize=(7, 5))
plt.hist(df["Average_Marks"], bins=5, edgecolor="black", color="skyblue")
plt.title("Distribution of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()
```

## Question 16: Box Plot

```python
plt.figure(figsize=(7, 5))
plt.boxplot([df["Python_Marks"], df["AI_Marks"], df["Network_Marks"]],
            labels=["Python", "AI", "Network"])
plt.title("Distribution of Marks Across Subjects")
plt.ylabel("Marks")
plt.show()
```

**Observation:** The box plot shows the median, spread (interquartile range) and range of marks for each subject. AI marks have the highest median, while Python marks show the widest spread, with the lowest marks at the lower whisker.

## Question 17: Pie Chart

```python
dept_counts = df["Department"].value_counts()
print(dept_counts)

plt.figure(figsize=(6, 6))
plt.pie(dept_counts, labels=dept_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Department-wise Distribution of Students")
plt.show()
```

# Part F: Integrated Application Problem
## Question 18: Student Performance Analysis System

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Read the dataset and display basic information
data = pd.read_csv("student_performance.csv")
print("Basic information:")
data.info()
print("\nShape:", data.shape)
```

```python
# 2. Total_Marks and Average_Marks
subj = ["Python_Marks", "AI_Marks", "Network_Marks"]
data["Total_Marks"] = data[subj].sum(axis=1)
data["Average_Marks"] = data[subj].mean(axis=1)
display(data[["Student_ID", "Name", "Total_Marks", "Average_Marks"]])
```

```python
# 3. Student with the highest Average_Marks
top = data.loc[data["Average_Marks"].idxmax()]
print("Highest Average_Marks:", top["Name"], "(Student_ID", top["Student_ID"], ") -", round(top["Average_Marks"], 2))
```

```python
# 4. Average marks for each department
dept_avg = data.groupby("Department")[subj + ["Average_Marks"]].mean()
display(dept_avg)
```

```python
# 5. Attendance >= 85 and Average_Marks >= 75
display(data[(data["Attendance"] >= 85) & (data["Average_Marks"] >= 75)])
```

```python
# 6. Sort by Average_Marks in descending order
sorted_data = data.sort_values("Average_Marks", ascending=False)
display(sorted_data)
```

```python
# 7. Bar chart: average marks of each department
plt.figure(figsize=(7, 5))
plt.bar(dept_avg.index, dept_avg["Average_Marks"], color=["steelblue", "orange", "green"])
plt.title("Department-wise Average Marks")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()
```

```python
# 8. Histogram: distribution of Average_Marks
plt.figure(figsize=(7, 5))
plt.hist(data["Average_Marks"], bins=5, edgecolor="black", color="skyblue")
plt.title("Distribution of Average Marks")
plt.xlabel("Average Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
# 9. Final DataFrame
final_df = data[["Student_ID", "Name", "Department", "Total_Marks", "Average_Marks"]]
display(final_df)
```
