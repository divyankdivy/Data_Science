## Question 1: Correlation-Based Feature Redundancy

Calculate the correlation matrix and identify feature pairs with positive correlation greater than 0.90.

```python
import pandas as pd

data = {
    'F1': [10, 20, 30, 40, 50],
    'F2': [12, 24, 36, 48, 60],
    'F3': [50, 40, 30, 20, 10],
    'F4': [15, 25, 20, 35, 30]
}

df = pd.DataFrame(data)
corr_matrix = df.corr()

print("Correlation Matrix:")
display(corr_matrix)

threshold = 0.90
redundant_pairs = []
for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        corr_value = corr_matrix.iloc[i, j]
        if corr_value > threshold:
            redundant_pairs.append((corr_matrix.columns[i],
                                    corr_matrix.columns[j],
                                    corr_value))

print("Feature pairs with positive correlation > 0.90:")
for f1, f2, value in redundant_pairs:
    print(f"{f1} and {f2}: {value:.4f}")
```

**Observation:** F1 and F2 have a positive correlation of 1.00, so they are redundant under the given threshold. The negative correlation between F1 and F3 does not meet the requirement for positive correlation.

## Question 2: Distance-Based Feature Similarity

Calculate Euclidean distance, Manhattan distance, and cosine similarity between the two feature vectors.

```python
import numpy as np

F1 = np.array([2, 4, 6, 8, 10])
F2 = np.array([3, 5, 7, 9, 11])

euclidean_distance = np.linalg.norm(F1 - F2)
manhattan_distance = np.sum(np.abs(F1 - F2))
cosine_similarity = np.dot(F1, F2) / (np.linalg.norm(F1) * np.linalg.norm(F2))

print("Euclidean Distance:", euclidean_distance)
print("Manhattan Distance:", manhattan_distance)
print("Cosine Similarity:", cosine_similarity)
```

**Observation:** The vectors are very similar in direction, as indicated by cosine similarity close to 1. Their Euclidean and Manhattan distances are non-zero because each corresponding element differs by 1.

## Question 4: Measuring Redundancy Between Binary Features

Calculate Hamming distance, n11, n00, n10, n01, Jaccard similarity, Jaccard distance, and SMC.

```python
import numpy as np

F1 = np.array([1, 1, 0, 1, 0, 0, 1, 0])
F2 = np.array([1, 0, 0, 1, 1, 0, 1, 0])

n11 = np.sum((F1 == 1) & (F2 == 1))
n00 = np.sum((F1 == 0) & (F2 == 0))
n10 = np.sum((F1 == 1) & (F2 == 0))
n01 = np.sum((F1 == 0) & (F2 == 1))

hamming_distance = np.sum(F1 != F2)
jaccard_similarity = n11 / (n11 + n10 + n01)
jaccard_distance = 1 - jaccard_similarity
smc = (n11 + n00) / (n11 + n10 + n01 + n00)

print("n11:", n11)
print("n00:", n00)
print("n10:", n10)
print("n01:", n01)
print("Hamming Distance:", hamming_distance)
print("Jaccard Similarity:", jaccard_similarity)
print("Jaccard Distance:", jaccard_distance)
print("Simple Matching Coefficient (SMC):", smc)
```

**Observation:** The vectors differ at 2 positions. Jaccard similarity considers shared 1s and ignores shared 0s, while SMC counts both shared 1s and shared 0s.

## Question 5: Comparing Binary Similarity Measures

Calculate Hamming distance, Jaccard distance, and SMC, and compare what each measure represents.

```python
import numpy as np

F1 = np.array([1, 0, 1, 1, 0, 1, 0, 0])
F2 = np.array([1, 0, 0, 1, 1, 1, 0, 1])

n11 = np.sum((F1 == 1) & (F2 == 1))
n00 = np.sum((F1 == 0) & (F2 == 0))
n10 = np.sum((F1 == 1) & (F2 == 0))
n01 = np.sum((F1 == 0) & (F2 == 1))

hamming_distance = np.sum(F1 != F2)
jaccard_similarity = n11 / (n11 + n10 + n01)
jaccard_distance = 1 - jaccard_similarity
smc = (n11 + n00) / len(F1)

print("Hamming Distance:", hamming_distance)
print("Jaccard Distance:", jaccard_distance)
print("Simple Matching Coefficient (SMC):", smc)
```

**Observation:** Hamming distance counts mismatched positions (3). Jaccard distance measures dissimilarity based only on positions where at least one feature is 1 (0.50). SMC measures the proportion of all matching positions, including shared zeros (0.625).

## Question 6: Identifying Redundant Features

Calculate the correlation matrix, identify pairs with correlation greater than 0.95, and create a DataFrame after removing redundant features.

```python
import pandas as pd

data = {
    'Study_Hours': [2, 4, 6, 8, 10, 12],
    'Practice_Hours': [3, 6, 9, 12, 15, 18],
    'Sleep_Hours': [8, 7, 8, 6, 7, 8],
    'Screen_Time': [2, 3, 4, 5, 6, 7]
}

df = pd.DataFrame(data)
corr_matrix = df.corr()

print("Correlation Matrix:")
display(corr_matrix)

threshold = 0.95
redundant_pairs = []
for i in range(len(corr_matrix.columns)):
    for j in range(i + 1, len(corr_matrix.columns)):
        value = corr_matrix.iloc[i, j]
        if value > threshold:
            redundant_pairs.append((corr_matrix.columns[i],
                                    corr_matrix.columns[j], value))

print("Redundant feature pairs (correlation > 0.95):")
for f1, f2, value in redundant_pairs:
    print(f"{f1} and {f2}: {value:.4f}")

# Keep Study_Hours as the representative of the perfectly correlated
# increasing features; remove Practice_Hours and Screen_Time.
reduced_df = df.drop(columns=['Practice_Hours', 'Screen_Time'])

print("\nDataFrame after removing redundant features:")
display(reduced_df)
```

**Observation:** Study_Hours, Practice_Hours, and Screen_Time are perfectly positively correlated with one another. Keeping Study_Hours as the representative feature and removing Practice_Hours and Screen_Time avoids retaining duplicate linear information. Sleep_Hours is retained.

## Question 8: Finding the Most Similar Feature Pair

Each row represents a feature. Calculate cosine similarity between every pair, display the similarity matrix, and identify the highest-similarity pair.

```python
import numpy as np
import pandas as pd

data = np.array([
    [2, 4, 6, 8, 10],
    [3, 6, 9, 12, 15],
    [10, 8, 6, 4, 2],
    [4, 5, 7, 8, 9]
])

feature_names = ['F1', 'F2', 'F3', 'F4']
norms = np.linalg.norm(data, axis=1)
similarity_matrix = (data @ data.T) / np.outer(norms, norms)

similarity_df = pd.DataFrame(similarity_matrix,
                             index=feature_names,
                             columns=feature_names)

print("Cosine Similarity Matrix:")
display(similarity_df)

best_pair = None
best_similarity = -np.inf
for i in range(len(feature_names)):
    for j in range(i + 1, len(feature_names)):
        if similarity_matrix[i, j] > best_similarity:
            best_similarity = similarity_matrix[i, j]
            best_pair = (feature_names[i], feature_names[j])

print("Most similar pair:", best_pair)
print("Highest cosine similarity:", best_similarity)
```

**Observation:** F1 and F2 have the highest cosine similarity of 1.0. F2 is a positive scalar multiple of F1, so the two vectors point in exactly the same direction.

## Question 13: Comparing Multiple Binary Features

Calculate Hamming distances for F1–F2 and F1–F3, compare all feature pairs to identify the most similar pair, and explain a distance of zero.

```python
import pandas as pd
import numpy as np

data = {
    'F1': [1, 1, 0, 1, 0, 1],
    'F2': [1, 1, 0, 1, 0, 1],
    'F3': [0, 1, 1, 0, 1, 0],
    'F4': [1, 0, 1, 0, 1, 1]
}

df = pd.DataFrame(data)

d_f1_f2 = np.sum(df['F1'].to_numpy() != df['F2'].to_numpy())
d_f1_f3 = np.sum(df['F1'].to_numpy() != df['F3'].to_numpy())

print("Hamming distance between F1 and F2:", d_f1_f2)
print("Hamming distance between F1 and F3:", d_f1_f3)

# Calculate all unique pairwise Hamming distances.
features = df.columns
distance_pairs = []
for i in range(len(features)):
    for j in range(i + 1, len(features)):
        distance = np.sum(df[features[i]].to_numpy() != df[features[j]].to_numpy())
        distance_pairs.append((features[i], features[j], int(distance)))

print("\nAll pairwise Hamming distances:")
for f1, f2, distance in distance_pairs:
    print(f"{f1} and {f2}: {distance}")

minimum_distance = min(item[2] for item in distance_pairs)
most_similar_pairs = [(f1, f2) for f1, f2, d in distance_pairs
                      if d == minimum_distance]

print("\nMost similar pair(s):", most_similar_pairs)
print("Minimum Hamming distance:", minimum_distance)
```

**Observation:** F1 and F2 have Hamming distance zero and are identical. A Hamming distance of zero means that the two vectors have equal values at every position.

## Question 14: Calculating Minkowski Distance

Calculate Minkowski distance for r = 1, r = 2, and r = 3. Identify the Manhattan and Euclidean cases.

```python
import numpy as np

F1 = np.array([2, 4, 6, 8])
F2 = np.array([5, 6, 8, 10])

def minkowski_distance(x, y, r):
    return np.sum(np.abs(x - y) ** r) ** (1 / r)

for r in [1, 2, 3]:
    print(f"Minkowski distance (r={r}):", minkowski_distance(F1, F2, r))

print("r = 1 gives Manhattan distance.")
print("r = 2 gives Euclidean distance.")
```

**Observation:** For r = 1, the Minkowski distance is the Manhattan distance (9). For r = 2, it is the Euclidean distance (approximately 4.5826). The r = 3 distance is approximately 3.7084.

## Question 15: Feature Similarity Using Cosine Similarity

Calculate cosine similarity between F1 and F2, and between F1 and F3. Identify the greater similarity.

```python
import numpy as np

F1 = np.array([1, 2, 3, 4])
F2 = np.array([2, 4, 6, 8])
F3 = np.array([4, 3, 2, 1])

def cosine_similarity(x, y):
    return np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))

sim_f1_f2 = cosine_similarity(F1, F2)
sim_f1_f3 = cosine_similarity(F1, F3)

print("Cosine similarity (F1, F2):", sim_f1_f2)
print("Cosine similarity (F1, F3):", sim_f1_f3)

if sim_f1_f2 > sim_f1_f3:
    print("F1 and F2 have greater cosine similarity.")
else:
    print("F1 and F3 have greater cosine similarity.")
```

**Observation:** F1 and F2 have cosine similarity 1.0 because F2 is a positive scalar multiple of F1. Their similarity is greater than that of F1 and F3 (approximately 0.6667).

## Question 16: Identifying Redundancy in a Dataset

Calculate the correlation matrix, identify pairs with absolute correlation equal to 1, remove redundant features, and display the resulting DataFrame.

```python
import pandas as pd

data = {
    'Temperature': [20, 25, 30, 35, 40, 45],
    'Temp_Celsius': [20, 25, 30, 35, 40, 45],
    'Humidity': [60, 65, 55, 70, 50, 75],
    'Pressure': [1000, 1005, 1010, 1015, 1020, 1025]
}

df = pd.DataFrame(data)
corr_matrix = df.corr()

print("Correlation Matrix:")
display(corr_matrix)

perfect_pairs = []
columns = corr_matrix.columns
for i in range(len(columns)):
    for j in range(i + 1, len(columns)):
        value = corr_matrix.iloc[i, j]
        if abs(value) == 1:
            perfect_pairs.append((columns[i], columns[j], value))

print("Feature pairs with absolute correlation equal to 1:")
for f1, f2, value in perfect_pairs:
    print(f"{f1} and {f2}: {value:.2f}")

# Temperature, Temp_Celsius, and Pressure are perfectly correlated.
# Keep Temperature as the representative and remove the other two.
reduced_df = df.drop(columns=['Temp_Celsius', 'Pressure'])

print("\nDataFrame after removing redundant features:")
display(reduced_df)
```

**Observation:** Temperature, Temp_Celsius, and Pressure are perfectly positively correlated. Keeping Temperature as the representative feature and removing Temp_Celsius and Pressure leaves Temperature and Humidity.
